#### **Outlier Detection**

```python
def detect_outliers_iqr(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Lọc ra các dòng là outlier
    outliers = dataframe[(dataframe[column] < lower_bound) | (dataframe[column] > upper_bound)]
    
    return len(outliers), lower_bound, upper_bound
```

```python
num_outliers, lower, upper = detect_outliers_iqr(data, 'Total Amount')
total_rows = len(data)
outlier_percent = (num_outliers / total_rows) * 100
```

```python
print(f"-> Biến mục tiêu: Total Amount")
print(f"   - Ngưỡng dưới an toàn (Lower Bound): {lower:.2f}")
print(f"   - Ngưỡng trên an toàn (Upper Bound): {upper:.2f}")
print(f"   - Số lượng hóa đơn Ngoại lai (Outliers): {num_outliers:,} / {total_rows:,} hóa đơn")
print(f"   - Tỷ lệ Ngoại lai: {outlier_percent:.2f}%")
```

### **Data Preprocessing**

```python
from sklearn.model_selection import train_test_split
```

```python
data.head()
```

```python
train_df, test_df = train_test_split(data, test_size=0.2, random_state=42)
```

```python
print(train_df.shape)
print(test_df.shape)
```

```python
def apply_stateless_cleaning(data):
    df_clean = data.copy()
    
    cols_to_drop = ['Transaction ID', 'Customer ID']
    df_clean = df_clean.drop(columns=[c for c in cols_to_drop if c in df_clean.columns])
    
    df_clean['Date'] = pd.to_datetime(df_clean['Date'], errors='coerce')
    df_clean['Gender'] = df_clean['Gender'].astype('category')
    
    num_cols = ['Age', 'Quantity', 'Price per Unit', 'Total Amount']
    for col in num_cols:
        df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')
        
    return df_clean
```

```python
train_df = apply_stateless_cleaning(train_df)
test_df = apply_stateless_cleaning(test_df)
```

```python
train_df = train_df.drop_duplicates(keep='first')

train_df = train_df[(train_df['Age'] >= 10) & (train_df['Price per Unit'] >= 0)]

valid_categories = ['Beauty', 'Clothing', 'Electronics']
train_df = train_df[train_df['Product Category'].isin(valid_categories)]

train_df = train_df.dropna()
```

```python
train_df.shape
```

```python
X_train_clean = train_df.drop(columns=['Total Amount'])
y_train = train_df['Total Amount']

X_test_clean = test_df.drop(columns=['Total Amount'])
y_test = test_df['Total Amount']
```

```python
X_train_clean.to_csv('./data/preprocessed-data/X_train_clean.csv', index=False)
X_test_clean.to_csv('./data/preprocessed-data/X_test_clean.csv', index=False)
y_train.to_csv('./data/preprocessed-data/y_train.csv', index=False)
y_test.to_csv('./data/preprocessed-data/y_test.csv', index=False)
```
